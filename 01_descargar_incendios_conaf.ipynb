# Descargar Dataset de Incendios Forestales CONAF (2010-2020)

Este notebook descarga el dataset **"Registro histórico de incendios forestales"** de la Plataforma de Datos para la Resiliencia ante Desastres (PyDataverse) y lo prepara para cruzar con variables climáticas ERA5.

## 1. Instalar PyDataverse

```python
import subprocess
import sys

try:
    import pyDataverse
    print("✓ PyDataverse ya está instalado")
except ImportError:
    print("Instalando PyDataverse 0.3.3...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "pyDataverse==0.3.3"])
    print("✓ PyDataverse instalado correctamente")
```

## 2. Importar librerías

```python
import os
import json
from pathlib import Path

import pandas as pd
import numpy as np

from pyDataverse.api import NativeApi, DataAccessApi

print("✓ Librerías importadas correctamente")
```

## 3. Configurar credenciales

Al correr la celda de abajo te va a pedir la clave por teclado (no se muestra en pantalla).
Consíguela gratis en https://www.plataformadedatos.cl/user/developer -- no se guarda en ningún archivo.

```python
import getpass

# La clave NUNCA se hardcodea (antes había una API key real en texto plano acá, lista
# para quedar expuesta en el historial de este repo público). Se pide por teclado, oculta,
# cada vez que se corre el notebook -- funciona igual en Colab, Jupyter local, o donde sea.
API_KEY = getpass.getpass(
    "Pega tu API key de https://www.plataformadedatos.cl/user/developer: "
)
BASE_URL = "https://datospararesiliencia.cl"
DATASET_ID = "doi:10.71578/UXAUN5"
DOWNLOAD_DIR = "./datos_incendios"
Path(DOWNLOAD_DIR).mkdir(exist_ok=True)

print(f"\u2713 Configuraci\u00f3n lista")
print(f"  Base URL: {BASE_URL}")
print(f"  Dataset ID: {DATASET_ID}")
print(f"  Directorio: {DOWNLOAD_DIR}")
```

## 4. Conectar a PyDataverse

```python
try:
    api = NativeApi(BASE_URL, API_KEY)
    data_api = DataAccessApi(BASE_URL, API_KEY)
    print("✓ Conexión exitosa a PyDataverse")
except Exception as e:
    print(f"✗ Error de conexión: {e}")
    print("Verifica tu API_KEY")
```

## 5. Obtener dataset

```python
dataset = api.get_dataset(DATASET_ID)
dataset_json = dataset.json()
print("✓ Dataset obtenido")
```

## 6. Listar archivos disponibles

```python
files_list = dataset_json['data']['latestVersion']['files']

print(f"\nArchivos disponibles ({len(files_list)} total):\n")
for idx, file in enumerate(files_list, 1):
    filename = file["dataFile"]["filename"]
    print(f"{idx:2d}. {filename}")
```

## 7. Descargar archivos CSV

```python
csv_files = [f for f in files_list if f["dataFile"]["filename"].endswith('.csv')]

print(f"Descargando {len(csv_files)} archivo(s)...\n")

for file in csv_files:
    filename = file["dataFile"]["filename"]
    file_id = file["dataFile"]["id"]
    filepath = os.path.join(DOWNLOAD_DIR, filename)
    
    print(f"Descargando: {filename}...", end="", flush=True)
    response = data_api.get_datafile(file_id, is_pid=False)
    
    with open(filepath, "wb") as f:
        f.write(response.content)
    
    file_size_mb = os.path.getsize(filepath) / (1024 * 1024)
    print(f" ✓ ({file_size_mb:.2f} MB)")

print("\n✓ Descarga completada")
```

## 8. Combinar CSV por temporada

Une los 18 archivos anuales (`YYYYYYYY.csv`, separador `|`) en un único dataset de eventos individuales y lo guarda como `incendios_conaf_raw.csv`. Se excluye `- Archivo Indice.csv`, que es solo un resumen por temporada sin fechas ni coordenadas.

```python
# Combinar los CSV por temporada (excluir "- Archivo Indice.csv", que es solo un resumen)
season_files = sorted(
    f for f in os.listdir(DOWNLOAD_DIR)
    if f.endswith('.csv') and f[:8].isdigit()      # p.ej. 20102011.csv
)
print(f"Combinando {len(season_files)} archivos de temporada...\n")

dfs = []
for f in season_files:
    df = pd.read_csv(os.path.join(DOWNLOAD_DIR, f), sep='|', low_memory=False)
    df['archivo_origen'] = f
    dfs.append(df)
    print(f"  {f}: {len(df):>6} filas")

incendios = pd.concat(dfs, ignore_index=True)
print(f"\n✓ Dataset combinado: {incendios.shape[0]} filas × {incendios.shape[1]} columnas")

# Guardar el crudo combinado para el notebook 02
raw_combinado = os.path.join(DOWNLOAD_DIR, "incendios_conaf_raw.csv")
incendios.to_csv(raw_combinado, index=False)
print(f"✓ Guardado: {raw_combinado} ({os.path.getsize(raw_combinado) / (1024*1024):.2f} MB)")
```
